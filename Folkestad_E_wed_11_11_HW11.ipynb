1. Write a ``Particle`` class that can be used to represent a particle with a mass, a 3-d position, and a 3-d velocity.

2. Write a method that can be used to compute the kinetic energy of the particle

3. Write a method that takes another particle as an argument and finds the distance between the two particles

4. Write a method that given a time interval ``dt`` will update the position of the particle to the new position based on the current position and velocity.

5. Write a ``ChargedParticle`` class that inherits from the ``Particle`` class, but also has an attribute for the charge of the particle.


```python
import numpy as np

class Particle(object):

    def __init__(self, mass, position, velocity):
        self.mass = mass
        self.position = np.array(position)
        self.velocity = np.array(velocity)

    def kinetic_energy(self):
        return 0.5 * self.mass * np.sum(self.velocity**2)

    def distance(self, other):
        return np.sqrt(np.sum((self.position - other.position)**2))

    def update_position(self, dt):
        self.position = self.position + self.velocity * dt


class ChargedParticle(Particle):

    def __init__(self, mass, position, velocity, charge):
        super().__init__(mass, position, velocity)
        self.charge = charge


p1 = Particle(2, [0, 0, 0], [3, 0, 0])
p2 = Particle(2, [3, 4, 0], [0, 2, 0])

print("Kinetic energy:", p1.kinetic_energy())
print("Distance:", p1.distance(p2))

p1.update_position(2)
print("New position:", p1.position)

p3 = ChargedParticle(1, [0, 0, 0], [2, 0, 0], 1.6)

print("Charged particle mass:", p3.mass)
print("Charged particle charge:", p3.charge)
print("Charged particle kinetic energy:", p3.kinetic_energy())
```

```python

import numpy as np

class Particle(object):

    def __init__(self, mass, position, velocity):
        self.mass = mass
        self.position = np.array(position)
        self.velocity = np.array(velocity)

    def kinetic_energy(self):
        return 0.5 * self.mass * np.sum(self.velocity**2)

    def distance(self, other):
        return np.sqrt(np.sum((self.position - other.position)**2))

    def update_position(self, dt):
        self.position = self.position + self.velocity * dt


class ChargedParticle(Particle):

    def __init__(self, mass, position, velocity, charge):
        super().__init__(mass, position, velocity)
        self.charge = charge

    # Exercise 2: Interaction
    def interaction(self, other):

        if self.distance(other) < 0.25:

            # Avoid repeating an interaction after merging
            if self.charge == 0 and other.charge == 0:
                return

            print("interaction")

            # Opposite charges merge
            if self.charge * other.charge < 0:

                self.velocity = np.zeros(3)
                other.velocity = np.zeros(3)

                self.charge = 0
                other.charge = 0

                print("merge")

            # Same-sign charges repel
            else:

                self.velocity *= (
                    -1 * (self.charge + other.charge)
                    * (self.mass / other.mass)
                )

                other.velocity *= (
                    -1 * (self.charge + other.charge)
                    * (other.mass / self.mass)
                )

                print("repel")



p1 = ChargedParticle(1, [-5,-5,-5], [1,1,1], 0.5)
p2 = ChargedParticle(1, [5,5,5], [-1,-1,-1], -0.5)

dt = 0.05

for i in range(300):

    p1.update_position(dt)
    p2.update_position(dt)

    p1.interaction(p2)

    print("Step:", i)
    print("P1 position:", p1.position)
    print("P1 velocity:", p1.velocity)

    print("P2 position:", p2.position)
    print("P2 velocity:", p2.velocity)



p1 = ChargedParticle(1, [-5,-5,-5], [2,2,2], 0.5)
p2 = ChargedParticle(1, [5,5,5], [-1,-1,-1], 2.0)

dt = 0.05

for i in range(300):

    p1.update_position(dt)
    p2.update_position(dt)

    p1.interaction(p2)

    print("Step:", i)
    print("P1 position:", p1.position)
    print("P1 velocity:", p1.velocity)

    print("P2 position:", p2.position)
    print("P2 velocity:", p2.velocity)
```
